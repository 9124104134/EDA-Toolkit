## 2.2 Handling Missing Values

We handle missing values using suitable statistical techniques.
A copy of the original dataset is maintained to preserve the raw data.

```python
df_clean = df.copy()

print("Original dataset shape:", df.shape)
print("Working dataset shape:", df_clean.shape)
```

```python
age_median = df_clean["Age"].median()

df_clean["Age"] = df_clean["Age"].fillna(age_median)

print("Missing Age values:", df_clean["Age"].isnull().sum())
```

```python
embarked_mode = df_clean["Embarked"].mode()[0]

df_clean["Embarked"] = df_clean["Embarked"].fillna(embarked_mode)

print("Missing Embarked values:", df_clean["Embarked"].isnull().sum())
```

```python
df_clean = df_clean.drop(columns=["Cabin"])

print("Cabin column removed.")
print("Updated dataset shape:", df_clean.shape)
print("Remaining missing values:")
print(df_clean.isnull().sum())

print("\nTotal missing values:", df_clean.isnull().sum().sum())
```

## 2.3 Duplicate Removal

Duplicate records can affect the accuracy of data analysis.

```python
duplicate_count = df_clean.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
df_clean = df_clean.drop_duplicates()

print("Dataset shape after removing duplicates:", df_clean.shape)
print("Remaining duplicates:", df_clean.duplicated().sum())
```

## 2.4 Data Type Validation

We inspect the data types and convert suitable columns
into categorical data types for efficient analysis.

```python
print(df_clean.dtypes)
categorical_columns = ["Survived", "Pclass", "Sex", "Embarked"]

for column in categorical_columns:
    df_clean[column] = df_clean[column].astype("category")

print("Updated Data Types:")
print(df_clean.dtypes)
```

```python
print("Categorical Columns:")
print(df_clean.select_dtypes(include=["category"]).columns.tolist())

print("\nFinal Dataset Shape:", df_clean.shape)
```

## 2.5 Outlier Detection Using IQR

The Interquartile Range (IQR) method is used to identify
potential outliers in numerical columns.

```python
numerical_columns = df_clean.select_dtypes(include=["int64", "float64"]).columns

print("Numerical columns:")
print(numerical_columns.tolist())
```

```python
for column in numerical_columns:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[column] < lower_bound) |
        (df_clean[column] > upper_bound)
    ]

    print(f"\nColumn: {column}")
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Outlier Count:", len(outliers))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

for column in numerical_columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df_clean[column])

    plt.title(f"Outlier Detection - {column}")
    plt.xlabel(column)
    plt.show()
```

# Part 4: Feature Engineering

## 4.1 Extracting Passenger Titles

We extract passenger titles such as Mr, Mrs, Miss and Master
from passenger names to create a new feature called Title.

```python
df_features = df_clean.copy()

df_features["Title"] = df_features["Name"].str.extract(
    r",\s*([^.]*)\."
)

print(df_features[["Name", "Title"]].head(10))
```

```python
print(df_features["Title"].value_counts())
```

## 4.2 Creating Family Size Feature

We combine the SibSp and Parch columns to calculate
the total family size of each passenger.

A new feature called FamilySize is created.

```python
df_features["FamilySize"] = (
    df_features["SibSp"] + df_features["Parch"] + 1
)

print(df_features[[
    "Name", "SibSp", "Parch", "FamilySize"
]].head(10))
```

```python
print(df_features["FamilySize"].value_counts().sort_index())
plt.figure(figsize=(9, 5))

sns.countplot(data=df_features, x="FamilySize")

plt.title("Distribution of Family Size")
plt.xlabel("Family Size")
plt.ylabel("Passenger Count")
plt.show()
```

## 4.3 Identifying Passengers Travelling Alone

We create a new feature called IsAlone to identify passengers
who travelled without any family members.

A value of 1 indicates travelling alone, while 0 indicates
travelling with family.

```python
df_features["IsAlone"] = (
    df_features["FamilySize"] == 1
).astype(int)

print(df_features[[
    "Name", "FamilySize", "IsAlone"
]].head(10))
```

```python
alone_counts = df_features["IsAlone"].value_counts()

print("Passenger Travelling Status:")
print(alone_counts)

print("\nPercentage Distribution:")
print(df_features["IsAlone"].value_counts(normalize=True) * 100)
plt.figure(figsize=(7, 5))

sns.countplot(data=df_features, x="IsAlone")

plt.title("Passengers Travelling Alone vs With Family")
plt.xlabel("Travelling Status (0 = With Family, 1 = Alone)")
plt.ylabel("Passenger Count")
plt.xticks([0, 1], ["With Family", "Alone"])
plt.show()
```

```python
alone_survival = df_features.assign(
    Survived=df_features["Survived"].astype(int)
).groupby("IsAlone", observed=True)["Survived"].mean() * 100

print("Survival Rate by Travelling Status:")
print(alone_survival)
```

## 4.4 Creating Age Groups

We categorize passengers into different age groups using
their Age values. This makes demographic analysis easier.

```python
df_features["AgeGroup"] = pd.cut(
    df_features["Age"],
    bins=[0, 12, 19, 60, float("inf")],
    labels=["Child", "Teenager", "Adult", "Senior"],
    right=True
)

print(df_features[["Age", "AgeGroup"]].head(15))
```

```python
print("Age Group Distribution:")
print(df_features["AgeGroup"].value_counts().sort_index())
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df_features,
    x="AgeGroup",
    order=["Child", "Teenager", "Adult", "Senior"]
)

plt.title("Passenger Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Passenger Count")
plt.show()
```

## 4.5 Creating Fare Categories

We categorize passenger ticket fares into Low, Medium and High
using quantile-based binning.

```python
df_features["FareCategory"] = pd.qcut(
    df_features["Fare"],
    q=3,
    labels=["Low", "Medium", "High"]
)

print(df_features[["Fare", "FareCategory"]].head(15))
```

```python
print("Fare Category Distribution:")
print(df_features["FareCategory"].value_counts())
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df_features,
    x="FareCategory",
    order=["Low", "Medium", "High"]
)

plt.title("Passenger Distribution by Fare Category")
plt.xlabel("Fare Category")
plt.ylabel("Passenger Count")
plt.show()
```

## Part 5: Building a Reusable EDA Toolkit

### 5.1 Dataset Overview

This function provides a quick overview of any dataset, including its dimensions, column names, data types, and memory usage.

```python
def dataset_overview(data):
    print("=" * 50)
    print("DATASET OVERVIEW")
    print("=" * 50)

    print("\nNumber of Rows:", data.shape[0])
    print("Number of Columns:", data.shape[1])

    print("\nColumn Names:")
    print(data.columns.tolist())

    print("\nData Types:")
    print(data.dtypes)

    print("\nDataset Information:")
    data.info()
```

```python
dataset_overview(df_features)
```

### 5.2 Missing Value Analysis

This function identifies missing values and calculates their percentage in each column.

```python
def missing_value_report(data):

    report = pd.DataFrame({
        "Missing Count": data.isnull().sum(),
        "Missing Percentage": (data.isnull().sum() / len(data)) * 100
    })

    report = report[report["Missing Count"] > 0]

    if report.empty:
        print("No missing values found!")
    else:
        display(report.sort_values(
            by="Missing Percentage",
            ascending=False
        ))
```

```python
missing_value_report(df_features)
```

### 5.3 Column Summary

This function separates numerical and categorical columns and generates descriptive summaries.

```python
def column_summary(data):

    numerical_columns = data.select_dtypes(
        include=["number"]
    ).columns.tolist()

    categorical_columns = data.select_dtypes(
        include=["object", "category", "bool"]
    ).columns.tolist()

    print("Numerical Columns:")
    print(numerical_columns)

    print("\nCategorical Columns:")
    print(categorical_columns)

    if numerical_columns:
        print("\nNumerical Summary:")
        display(data[numerical_columns].describe())
    else:
        print("\nNo numerical columns found.")

    if categorical_columns:
        print("\nCategorical Summary:")
        display(data[categorical_columns].describe())
    else:
        print("\nNo categorical columns found.")
```

```python
column_summary(df_features)
```

## Part 6: Automated Data Visualization

### 6.1 Numerical Column Distributions

This function automatically identifies numerical columns and generates histograms to understand their distributions.

```python
def plot_numerical_distributions(data):

    numerical_columns = data.select_dtypes(
        include=["number"]
    ).columns.tolist()

    if not numerical_columns:
        print("No numerical columns found!")
        return

    for column in numerical_columns:

        plt.figure(figsize=(7, 4))

        sns.histplot(
            data=data,
            x=column,
            kde=True
        )

        plt.title(f"Distribution of {column}")
        plt.xlabel(column)
        plt.ylabel("Frequency")
        plt.tight_layout()
        plt.show()
```

```python
plot_numerical_distributions(df_features)
```

### 6.2 Categorical Column Distributions

This function automatically generates count plots for categorical columns to understand the frequency of each category.

```python
def plot_categorical_distributions(data):

    categorical_columns = data.select_dtypes(
        include=["object", "category", "bool"]
    ).columns.tolist()

    if not categorical_columns:
        print("No categorical columns found!")
        return

    for column in categorical_columns:

        plt.figure(figsize=(8, 4))

        order = data[column].value_counts().index

        sns.countplot(
            data=data,
            x=column,
            order=order
        )

        plt.title(f"Distribution of {column}")
        plt.xlabel(column)
        plt.ylabel("Count")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()
```

```python
plot_categorical_distributions(df_features)
```

### 6.3 Automated Correlation Analysis

Correlation analysis helps identify the strength and direction of relationships between numerical variables.

This function generates a correlation matrix and visualizes it using a heatmap.

```python
def correlation_analysis(data, exclude_columns=None):

    if exclude_columns is None:
        exclude_columns = []

    numerical_data = data.select_dtypes(
        include=["number"]
    ).drop(columns=exclude_columns, errors="ignore")

    if numerical_data.shape[1] < 2:
        print("At least two numerical columns are required.")
        return

    correlation_matrix = numerical_data.corr()

    print("Correlation Matrix:")
    display(correlation_matrix)

    plt.figure(figsize=(10, 7))

    sns.heatmap(
        correlation_matrix,
        annot=True,
        cmap="coolwarm",
        fmt=".2f",
        linewidths=0.5
    )

    plt.title("Correlation Heatmap")
    plt.tight_layout()
    plt.show()
```

```python
correlation_analysis(
    df_features,
    exclude_columns=["PassengerId"]
)
```
